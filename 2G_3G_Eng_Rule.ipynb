```python
import pandas as pd

# Raw GitHub file URL
url = "https://raw.githubusercontent.com/sathkumara-smna/258739K/main/Site%20Database%20from%20Sey.xlsx"

# Load Excel file
df = pd.read_excel(url)

# Show first few rows
df.head(2)
```

```python
df.columns = [
    '#',
    'Site_ID',
    'Site_Name',
    'RRU_2G',
    'RRU_3G',
    'RRU_4G',
    'AAU_5G',
    'Col_H',
    'Col_I',
    'Boards_4G',
    'Boards_5G',
    'BBU5900',
    'BBU3900',
    'BBU3910'
]
```

```python
# Create a new dataframe with only unique Site_IDs

df2g3g = df[['Site_ID']].drop_duplicates()

# Reset index
df2g3g = df2g3g.reset_index(drop=True)

# Show first rows
df2g3g.head(2)
```

```python
df2g3g['rru_count_2g'] = 0
df2g3g['rru_2g_bp'] = 150
df2g3g['power_2g'] = 0
df2g3g['rru_count_3g'] = 0
df2g3g['rru_3g_bp'] = 200
df2g3g['power_3g'] = 0
df2g3g['total_2g_3g_power'] = 0
# Show first rows
df2g3g.head(2)
```

```python
# Fill rru_count_2g from df using Site_ID matching
df2g3g['rru_count_2g'] = (
        df2g3g['Site_ID'].map(
        df.set_index('Site_ID')['RRU_2G']
    )
)
# Show first rows
df2g3g.head()
```

```python
# Calculate 2G power

df2g3g['power_2g'] = (
    df2g3g['rru_count_2g'] * df2g3g['rru_2g_bp']
)
# Show first rows
df2g3g.head(2)
```

```python
# Fill rru_count_3g from df using Site_ID matching

df2g3g['rru_count_3g'] = (
        df2g3g['Site_ID'].map(
        df.set_index('Site_ID')['RRU_3G']
    )
)
# Show first rows
df2g3g.head()
```

```python
# Calculate 3G power

df2g3g['power_3g'] = (df2g3g['rru_count_3g'] * df2g3g['rru_3g_bp'])
# Show first rows
df2g3g.head(2)
```

```python
# Show row for Site_ID = 163

df2g3g[df2g3g['Site_ID'] == 163]
```

```python
print(df2g3g.columns)
```

```python
# Calculate total 2G + 3G power

df2g3g['total_2g_3g_power'] = (
    df2g3g['power_2g'] + df2g3g['power_3g']
)
# Show first rows
df2g3g.head()
```

```python
df2g3g.to_excel("site_power_2g3g.xlsx", index=False)
```
